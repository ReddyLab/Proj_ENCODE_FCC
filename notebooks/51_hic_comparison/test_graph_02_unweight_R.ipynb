```r
# -----------------------------
# 0) Pick dataset + pull inputs
# -----------------------------
idx <- "fcc_astarr_macs_input_overlap"   # or "fcc_astarr_macs_input_union"

dat_edges <- lst_dat_region_annot_hic_import[[idx]]           # Hi-C edges
dat_nodes_raw <- lst_dat_region_annot_fcc_import[[idx]]       # FCC labels
dat_edges_lab <- lst_dat_region_annot_label[[idx]]            # edges + PP/PD/DD + AA/AI/...

# Deduplicate FCC labels to 1 row per Region
dat_nodes <- dat_nodes_raw %>% dplyr::distinct(Region, .keep_all = TRUE)

# -----------------------------
# 1) Build graph (UNDIRECTED)
#    UNWEIGHTED paths (hop count)
# -----------------------------
g <- igraph::graph_from_data_frame(
  d = dat_edges %>% dplyr::select(Region_1, Region_2),
  directed = FALSE
)
vcount(g); ecount(g)  # quick sanity

# -----------------------------
# 2) Node attributes table
# -----------------------------
g_tbl <- tidygraph::as_tbl_graph(g) %>%
  tidygraph::activate(nodes) %>%
  dplyr::mutate(name = name) %>%
  tibble::as_tibble()

node_df <- g_tbl %>%
  dplyr::select(name) %>%
  dplyr::left_join(
    dat_nodes %>% dplyr::select(Region, TSS_Proximity, Direction_Assay, Group),
    by = c("name" = "Region")
  ) %>%
  dplyr::mutate(ClassPD = paste(TSS_Proximity, Direction_Assay, sep = "|"))

stopifnot(sum(is.na(node_df$Group)) == 0)  # ensure we have labels for all graph nodes

# -----------------------------
# 3) Components & filter
# -----------------------------
comp <- igraph::components(g)
comp_sizes <- tibble::tibble(comp_id = seq_along(comp$csize), size = comp$csize)

min_size <- 50     # tweak as needed (25/50/100)
keep_ids <- comp_sizes %>% dplyr::filter(size >= min_size) %>% dplyr::pull(comp_id)

# map nodes to their component (for coverage QC)
node_df$component_id <- comp$membership[ match(node_df$name, igraph::V(g)$name) ]
kept_nodes <- node_df %>% dplyr::filter(component_id %in% keep_ids)

coverage_tbl <- node_df %>%
  dplyr::count(Group, name = "n_all") %>%
  dplyr::left_join(kept_nodes %>% dplyr::count(Group, name = "n_kept"), by = "Group") %>%
  dplyr::mutate(n_kept = tidyr::replace_na(n_kept, 0L),
                frac_kept = n_kept / n_all) %>%
  dplyr::arrange(frac_kept)
coverage_tbl

# =============================
# Chunk 4–5: helpers + config + run (UNWEIGHTED)
# =============================

## ---- Per-component class–class summary (UNWEIGHTED by default) ----
comp_pair_summary <- function(g_comp,
                              node_tbl_comp,
                              class_a, class_b,
                              class_col = "Group",
                              n_per_class = Inf,
                              weights_attr = NULL,   # NULL => unweighted hop count
                              sample_k = 10000,
                              drop_self_pairs = TRUE) {

  stopifnot("name" %in% names(node_tbl_comp), class_col %in% names(node_tbl_comp))

  A <- node_tbl_comp %>%
    dplyr::filter(.data[[class_col]] == class_a) %>%
    dplyr::pull(name)
  B <- node_tbl_comp %>%
    dplyr::filter(.data[[class_col]] == class_b) %>%
    dplyr::pull(name)

  if (is.finite(n_per_class)) {
    if (length(A) > n_per_class) A <- sample(A, n_per_class)
    if (length(B) > n_per_class) B <- sample(B, n_per_class)
  }

  present <- igraph::V(g_comp)$name
  A <- intersect(A, present)
  B <- intersect(B, present)

  if (length(A) == 0L || length(B) == 0L) {
    return(tibble::tibble(
      class_a, class_b,
      comp_size = igraph::vcount(g_comp),
      nA = length(A), nB = length(B),
      n_pairs = 0L, n_self_removed = 0L,
      frac_reachable = NA_real_,
      mean_dist = NA_real_, median_dist = NA_real_,
      q25 = NA_real_, q75 = NA_real_,
      sample_dist = list(numeric(0))
    ))
  }

  # NULL => unweighted (hop count)
  w <- if (is.null(weights_attr)) NA else {
    if (igraph::ecount(g_comp) > 0 && weights_attr %in% igraph::edge_attr_names(g_comp)) {
      igraph::edge_attr(g_comp, weights_attr)
    } else NA
  }

  D <- igraph::distances(g_comp, v = A, to = B, weights = w, mode = "all")

  # remove same-node diagonal for within-class
  n_self_removed <- 0L
  if (drop_self_pairs && identical(class_a, class_b)) {
    ra <- rownames(D); cb <- colnames(D)
    common <- intersect(ra, cb)
    if (length(common)) {
      D[cbind(match(common, ra), match(common, cb))] <- NA_real_
      n_self_removed <- length(common)
    }
  }

  Dfin <- as.numeric(D[is.finite(D)])
  total_pairs_nominal <- length(A) * length(B) - n_self_removed
  nfin <- length(Dfin)
  frac_reach <- if (total_pairs_nominal > 0) nfin / total_pairs_nominal else NA_real_

  k <- min(sample_k, nfin)
  Dsample <- if (k > 0) sample(Dfin, k) else numeric(0)

  tibble::tibble(
    class_a = class_a, class_b = class_b,
    comp_size = igraph::vcount(g_comp),
    nA = length(A), nB = length(B),
    n_pairs = nfin, n_self_removed = n_self_removed,
    frac_reachable = frac_reach,
    mean_dist = if (nfin) mean(Dfin) else NA_real_,
    median_dist = if (nfin) stats::median(Dfin) else NA_real_,
    q25 = if (nfin) stats::quantile(Dfin, 0.25) else NA_real_,
    q75 = if (nfin) stats::quantile(Dfin, 0.75) else NA_real_,
    sample_dist = list(Dsample)
  )
}

## ---- Union-of-classes runner (handles symmetric vs asymmetric) ----
run_pair_union <- function(g, comp, keep_ids, node_df2,
                           setA, setB,
                           class_col = "ClassPD",
                           min_n_A  = 5,    # asymmetric: min nodes in A
                           min_n_B  = 5,    # asymmetric: min nodes in B
                           min_n_sym = 2,   # symmetric: min nodes in class/union
                           weights_attr = NULL,  # NULL => unweighted hops
                           sample_k = 5000,
                           drop_self_pairs = TRUE,
                           label = NA_character_, grouping = NA_character_) {

  stopifnot(length(setA) > 0, length(setB) > 0)
  symmetric <- setequal(setA, setB)

  per_comp <- purrr::map_dfr(keep_ids, function(cid){
    verts <- igraph::V(g)[comp$membership == cid]
    gC    <- igraph::induced_subgraph(g, verts)
    if (igraph::vcount(gC) < 2L) return(tibble::tibble())

    nodeC <- node_df2 %>% dplyr::filter(name %in% igraph::V(gC)$name)
    if (!(class_col %in% names(nodeC))) return(tibble::tibble())

    if (symmetric) {
      # within a single class (AA / II / RR)
      n_sym <- sum(nodeC[[class_col]] %in% setA)
      if (n_sym < min_n_sym) return(tibble::tibble())

      if (length(setA) == 1L) {
        out <- comp_pair_summary(
          g_comp = gC, node_tbl_comp = nodeC,
          class_a = setA[[1]], class_b = setA[[1]],
          class_col = class_col,
          n_per_class = Inf,
          weights_attr = weights_attr,
          sample_k = sample_k,
          drop_self_pairs = drop_self_pairs
        )
      } else {
        # within union (rare): label them the same on both sides
        nodeC2 <- nodeC %>%
          dplyr::mutate(UnionLabel = ifelse(.data[[class_col]] %in% setA, "SET_U", NA_character_)) %>%
          dplyr::filter(!is.na(UnionLabel))

        out <- comp_pair_summary(
          g_comp = gC, node_tbl_comp = nodeC2,
          class_a = "SET_U", class_b = "SET_U",
          class_col = "UnionLabel",
          n_per_class = Inf,
          weights_attr = weights_attr,
          sample_k = sample_k,
          drop_self_pairs = drop_self_pairs
        )
      }

    } else {
      # across two unions (AI / AR / IR)
      nA <- sum(nodeC[[class_col]] %in% setA)
      nB <- sum(nodeC[[class_col]] %in% setB)
      if (nA < min_n_A || nB < min_n_B) return(tibble::tibble())

      nodeC2 <- nodeC %>%
        dplyr::mutate(UnionLabel = dplyr::case_when(
          .data[[class_col]] %in% setA ~ "SET_A",
          .data[[class_col]] %in% setB ~ "SET_B",
          TRUE ~ NA_character_
        )) %>%
        dplyr::filter(!is.na(UnionLabel))

      out <- comp_pair_summary(
        g_comp = gC, node_tbl_comp = nodeC2,
        class_a = "SET_A", class_b = "SET_B",
        class_col = "UnionLabel",
        n_per_class = Inf,
        weights_attr = weights_attr,
        sample_k = sample_k,
        drop_self_pairs = drop_self_pairs
      )
    }

    out %>% dplyr::mutate(component_id = cid, comp_size = igraph::vcount(gC))
  })

  per_comp %>%
    dplyr::summarize(
      comps_with_pairs = sum(n_pairs > 0),
      total_pairs      = sum(n_pairs),
      mean_dist        = if (sum(n_pairs) > 0)
                           stats::weighted.mean(mean_dist, w = n_pairs, na.rm = TRUE) else NA_real_,
      global_median    = { samp <- unlist(sample_dist, use.names = FALSE)
                           if (length(samp)) stats::median(samp) else NA_real_ },
      pooled_sample    = list(unlist(sample_dist, use.names = FALSE)),
      .groups = "drop"
    ) %>%
    dplyr::mutate(label = label, grouping = grouping)
}

## ---- Config for PP / PD / DD × {AA, AI, AR, II, IR, RR} ----
PA <- "Proximal|Active"; PI <- "Proximal|Inactive"; PR <- "Proximal|Repressive"
DA <- "Distal|Active";   DI <- "Distal|Inactive";   DR <- "Distal|Repressive"

cfg <- dplyr::bind_rows(
  # PP (proximal–proximal)
  tibble::tibble(
    grouping = "PP",
    label    = c("Active-Active","Active-Inactive","Active-Repressive",
                 "Inactive-Inactive","Inactive-Repressive","Repressive-Repressive"),
    A = list(c(PA), c(PA), c(PA), c(PI), c(PI), c(PR)),
    B = list(c(PA), c(PI), c(PR), c(PI), c(PR), c(PR))
  ),
  # DD (distal–distal)
  tibble::tibble(
    grouping = "DD",
    label    = c("Active-Active","Active-Inactive","Active-Repressive",
                 "Inactive-Inactive","Inactive-Repressive","Repressive-Repressive"),
    A = list(c(DA), c(DA), c(DA), c(DI), c(DI), c(DR)),
    B = list(c(DA), c(DI), c(DR), c(DI), c(DR), c(DR))
  ),
  # PD (proximal–distal; unions cover both directions)
  tibble::tibble(
    grouping = "PD",
    label    = c("Active-Active","Active-Inactive","Active-Repressive",
                 "Inactive-Inactive","Inactive-Repressive","Repressive-Repressive"),
    A = list(c(PA),       c(PA, PI),     c(PA, PR),     c(PI),       c(PI, PR),     c(PR)),
    B = list(c(DA),       c(DI, DA),     c(DR, DA),     c(DI),       c(DR, DI),     c(DR))
  )
) %>%
  dplyr::mutate(
    label    = factor(label,
                      levels = c("Active-Active","Active-Inactive","Active-Repressive",
                                 "Inactive-Inactive","Inactive-Repressive","Repressive-Repressive")),
    grouping = factor(grouping, levels = c("PP","PD","DD"))
  )

## ---- Run all bins (UNWEIGHTED) ----
node_df2 <- node_df %>% dplyr::mutate(ClassPD = paste(TSS_Proximity, Direction_Assay, sep="|"))

tri_res <- purrr::pmap_dfr(
  list(cfg$A, cfg$B, cfg$label, cfg$grouping),
  ~ run_pair_union(g, comp, keep_ids, node_df2,
                   setA = ..1, setB = ..2,
                   class_col = "ClassPD",
                   min_n_A = 2, min_n_B = 2, min_n_sym = 2,
                   weights_attr = NULL,     # unweighted hops
                   sample_k = 5000,
                   drop_self_pairs = TRUE,
                   label = ..3, grouping = ..4)
)

# quick look
tri_res %>%
  dplyr::select(grouping, label, comps_with_pairs, total_pairs, mean_dist, global_median) %>%
  print(n = Inf)
```

```r
# (Optional) quick Wilcoxon shifts vs II using pooled samples
get_samp <- function(grp, lab) {
  tri_res %>%
    dplyr::filter(grouping == grp, label == lab) %>%
    tidyr::unnest_longer(pooled_sample, values_to = "d") %>%
    dplyr::pull(d)
}
wilcox_delta <- function(grp, lab, cap = 10000) {
  x <- get_samp(grp, lab)
  y <- get_samp(grp, "Inactive-Inactive")
  if (length(x) > cap) x <- sample(x, cap)
  if (length(y) > cap) y <- sample(y, cap)
  data.frame(
    grouping = grp, label = lab,
    median_x = stats::median(x), median_ii = stats::median(y),
    shift = stats::median(x) - stats::median(y),
    p = stats::wilcox.test(x, y, alternative = "two.sided")$p.value
  )
}
bind_rows(
  wilcox_delta("PP","Active-Active"),
  wilcox_delta("PP","Repressive-Repressive"),
  wilcox_delta("DD","Active-Active"),
  wilcox_delta("DD","Repressive-Repressive"),
  wilcox_delta("PD","Active-Active"),
  wilcox_delta("PD","Repressive-Repressive")
) %>% tibble::as_tibble() %>% print(n = Inf)
```

```r
### Safe helper for class–class distances inside a component
# overwrite any previous definitions
comp_pair_summary <- function(g_comp,
                              node_tbl_comp,
                              class_a, class_b,
                              class_col = "Group",
                              n_per_class = Inf,
                              weights_attr = "len",   # set NULL for unweighted
                              sample_k = 10000,
                              drop_self_pairs = TRUE) {

  A <- node_tbl_comp %>% dplyr::filter(.data[[class_col]] == class_a) %>% dplyr::pull(name)
  B <- node_tbl_comp %>% dplyr::filter(.data[[class_col]] == class_b) %>% dplyr::pull(name)

  if (is.finite(n_per_class)) {
    if (length(A) > n_per_class) A <- sample(A, n_per_class)
    if (length(B) > n_per_class) B <- sample(B, n_per_class)
  }

  present <- igraph::V(g_comp)$name
  A <- intersect(A, present); B <- intersect(B, present)

  if (length(A) == 0 || length(B) == 0) {
    return(tibble::tibble(
      class_a, class_b,
      comp_size = igraph::vcount(g_comp),
      nA = length(A), nB = length(B),
      n_pairs = 0, n_self_removed = 0,
      frac_reachable = NA_real_,
      mean_dist = NA_real_, median_dist = NA_real_,
      q25 = NA_real_, q75 = NA_real_,
      sample_dist = list(numeric(0))
    ))
  }

  if (!is.null(weights_attr) &&
      igraph::ecount(g_comp) > 0 &&
      weights_attr %in% igraph::edge_attr_names(g_comp)) {
    w <- igraph::edge_attr(g_comp, weights_attr)
  } else {
    w <- NA  # unweighted hops
  }

  D <- igraph::distances(g_comp, v = A, to = B, weights = w, mode = "all")

  n_self_removed <- 0L
  if (drop_self_pairs && identical(class_a, class_b)) {
    ra <- rownames(D); cb <- colnames(D)
    common <- intersect(ra, cb)
    if (length(common)) {
      D[cbind(match(common, ra), match(common, cb))] <- NA_real_
      n_self_removed <- length(common)
    }
  }

  Dfin <- as.numeric(D[is.finite(D)])
  total_pairs_nominal <- length(A) * length(B) - n_self_removed
  nfin <- length(Dfin)
  frac_reach <- if (total_pairs_nominal > 0) nfin / total_pairs_nominal else NA_real_

  k <- min(sample_k, nfin)
  Dsample <- if (k > 0) sample(Dfin, k) else numeric(0)

  tibble::tibble(
    class_a = class_a, class_b = class_b,
    comp_size = igraph::vcount(g_comp),
    nA = length(A), nB = length(B),
    n_pairs = nfin, n_self_removed = n_self_removed,
    frac_reachable = frac_reach,
    mean_dist = if (nfin) mean(Dfin) else NA_real_,
    median_dist = if (nfin) stats::median(Dfin) else NA_real_,
    q25 = if (nfin) stats::quantile(Dfin, 0.25) else NA_real_,
    q75 = if (nfin) stats::quantile(Dfin, 0.75) else NA_real_,
    sample_dist = list(Dsample)
  )
}
```

```r
### 6) Shortest-path distance distributions per connection type
# shorthand labels
PA <- "Proximal|Active"; PI <- "Proximal|Inactive"; PR <- "Proximal|Repressive"
DA <- "Distal|Active";   DI <- "Distal|Inactive";   DR <- "Distal|Repressive"

run_pair_union <- function(g, comp, keep_ids, node_df2,
                           setA, setB,
                           class_col = "ClassPD",
                           min_n_A = 5,          # total nodes from setA in a component
                           min_n_B = 5,          # total nodes from setB in a component
                           weights_attr = "len",
                           sample_k = 5000,
                           drop_self_pairs = TRUE,
                           label = NA_character_, grouping = NA_character_) {

  purrr::map_dfr(keep_ids, function(cid){
    verts <- igraph::V(g)[comp$membership == cid]
    gC    <- igraph::induced_subgraph(g, verts)
    nodeC <- node_df2 %>% dplyr::filter(name %in% igraph::V(gC)$name)
    if (igraph::vcount(gC) < 2L) return(tibble::tibble())

    # ----- special case: same-class (e.g., PA vs PA) -----
    if (identical(setA, setB) && length(setA) == 1) {
      cls  <- setA[[1]]
      n_cls <- sum(nodeC[[class_col]] == cls)
      if (n_cls < max(min_n_A, min_n_B)) return(tibble::tibble())

      out <- comp_pair_summary(
        g_comp        = gC,
        node_tbl_comp = nodeC,
        class_a       = cls,
        class_b       = cls,
        class_col     = class_col,
        n_per_class   = Inf,
        weights_attr  = weights_attr,
        sample_k      = sample_k,
        drop_self_pairs = drop_self_pairs
      )
      return(out %>% dplyr::mutate(component_id = cid, comp_size = igraph::vcount(gC)))
    }
    # ------------------------------------------------------

    # UNION-wise presence (not per subclass)
    nA <- sum(nodeC[[class_col]] %in% setA)
    nB <- sum(nodeC[[class_col]] %in% setB)
    if (nA < min_n_A || nB < min_n_B) return(tibble::tibble())

    nodeC2 <- nodeC %>%
      dplyr::mutate(UnionLabel = dplyr::case_when(
        .data[[class_col]] %in% setA ~ "SET_A",
        .data[[class_col]] %in% setB ~ "SET_B",
        TRUE ~ NA_character_
      )) %>%
      dplyr::filter(!is.na(UnionLabel))

    out <- comp_pair_summary(
      g_comp = gC,
      node_tbl_comp = nodeC2,
      class_a = "SET_A",
      class_b = "SET_B",
      class_col = "UnionLabel",
      n_per_class = Inf,
      weights_attr = weights_attr,
      sample_k = sample_k,
      drop_self_pairs = drop_self_pairs
    )

    out %>% dplyr::mutate(component_id = cid, comp_size = igraph::vcount(gC))
  }) %>%
    dplyr::summarize(
      comps_with_pairs = sum(n_pairs > 0),
      total_pairs      = sum(n_pairs),
      mean_dist        = if (sum(n_pairs) > 0)
                           stats::weighted.mean(mean_dist, w = n_pairs, na.rm = TRUE) else NA_real_,
      global_median    = { samp <- unlist(sample_dist, use.names = FALSE)
                           if (length(samp)) stats::median(samp) else NA_real_ },
      pooled_sample    = list(unlist(sample_dist, use.names = FALSE)),
      .groups = "drop"
    ) %>%
    dplyr::mutate(label = label, grouping = grouping)
}

# run all bins
node_df2 <- node_df %>% mutate(ClassPD = paste(TSS_Proximity, Direction_Assay, sep="|"))

# 18-bin configuration: which union-of-classes goes on each side
cfg <- bind_rows(
  # PP (proximal–proximal)
  tibble(
    grouping = "PP",
    label    = c("Active-Active","Active-Inactive","Active-Repressive",
                 "Inactive-Inactive","Inactive-Repressive","Repressive-Repressive"),
    A = list(c(PA), c(PA), c(PA), c(PI), c(PI), c(PR)),
    B = list(c(PA), c(PI), c(PR), c(PI), c(PR), c(PR))
  ),
  # DD (distal–distal)
  tibble(
    grouping = "DD",
    label    = c("Active-Active","Active-Inactive","Active-Repressive",
                 "Inactive-Inactive","Inactive-Repressive","Repressive-Repressive"),
    A = list(c(DA), c(DA), c(DA), c(DI), c(DI), c(DR)),
    B = list(c(DA), c(DI), c(DR), c(DI), c(DR), c(DR))
  ),
  # PD (proximal–distal; union both directions)
  tibble(
    grouping = "PD",
    label    = c("Active-Active","Active-Inactive","Active-Repressive",
                 "Inactive-Inactive","Inactive-Repressive","Repressive-Repressive"),
    # A = proximal unions, B = distal unions
    A = list(c(PA),       c(PA, PI),     c(PA, PR),     c(PI),       c(PI, PR),     c(PR)),
    B = list(c(DA),       c(DI, DA),     c(DR, DA),     c(DI),       c(DR, DI),     c(DR))
  )
) %>%
  mutate(
    label    = factor(label, levels = c("Active-Active","Active-Inactive","Active-Repressive",
                                        "Inactive-Inactive","Inactive-Repressive","Repressive-Repressive")),
    grouping = factor(grouping, levels = c("PP","PD","DD"))
  )

# build results
tri_res <- pmap_dfr(
  list(cfg$A, cfg$B, cfg$label, cfg$grouping),
  ~ run_pair_union(g, comp, keep_ids, node_df2,
                   setA = ..1, setB = ..2,
                   class_col = "ClassPD",
                   min_n_A = 2, min_n_B = 2,
                   weights_attr = "len",
                   sample_k = 5000,
                   drop_self_pairs = TRUE,
                   label = ..3, grouping = ..4)
)
```

```r
# node_tbl must have columns: name, <class_col> (e.g., Group or TSS_Proximity)
comp_pair_summary <- function(g_comp, node_tbl_comp, class_a, class_b, class_col = "Group",
                              n_per_class = Inf) {

  A <- node_tbl_comp %>% filter(.data[[class_col]] == class_a) %>% pull(name)
  B <- node_tbl_comp %>% filter(.data[[class_col]] == class_b) %>% pull(name)

  # Optional subsampling to control runtime
  if (is.finite(n_per_class)) {
    if (length(A) > n_per_class) A <- sample(A, n_per_class)
    if (length(B) > n_per_class) B <- sample(B, n_per_class)
  }

  if (length(A) == 0 || length(B) == 0) {
    return(tibble(class_a, class_b, comp_size = vcount(g_comp),
                  nA = length(A), nB = length(B),
                  n_pairs = 0, frac_reachable = NA_real_,
                  mean_dist = NA_real_, median_dist = NA_real_,
                  q25 = NA_real_, q75 = NA_real_))
  }

  D <- distances(g_comp, v = A, to = B, weights = E(g_comp)$len, mode = "all")
  Dfin <- as.numeric(D[is.finite(D)])

  tibble(
    class_a = class_a,
    class_b = class_b,
    comp_size = vcount(g_comp),
    nA = length(A),
    nB = length(B),
    n_pairs = length(Dfin),
    frac_reachable = length(Dfin) / (length(A) * length(B)),
    mean_dist = if (length(Dfin)) mean(Dfin) else NA_real_,
    median_dist = if (length(Dfin)) median(Dfin) else NA_real_,
    q25 = if (length(Dfin)) quantile(Dfin, 0.25) else NA_real_,
    q75 = if (length(Dfin)) quantile(Dfin, 0.75) else NA_real_
  )
}

```

```r
rm(list = c("comp_pair_summary"))  # remove any old definition (safe if absent)

comp_pair_summary <- function(g_comp,
                              node_tbl_comp,
                              class_a, class_b,
                              class_col = "Group",
                              n_per_class = Inf,
                              weights_attr = "len",   # set NULL for unweighted
                              sample_k = 10000,
                              drop_self_pairs = TRUE) {

  A <- node_tbl_comp %>% dplyr::filter(.data[[class_col]] == class_a) %>% dplyr::pull(name)
  B <- node_tbl_comp %>% dplyr::filter(.data[[class_col]] == class_b) %>% dplyr::pull(name)

  if (is.finite(n_per_class)) {
    if (length(A) > n_per_class) A <- sample(A, n_per_class)
    if (length(B) > n_per_class) B <- sample(B, n_per_class)
  }

  present <- igraph::V(g_comp)$name
  A <- intersect(A, present)
  B <- intersect(B, present)

  if (length(A) == 0 || length(B) == 0) {
    return(tibble::tibble(
      class_a, class_b,
      comp_size = igraph::vcount(g_comp),
      nA = length(A), nB = length(B),
      n_pairs = 0, n_self_removed = 0,
      frac_reachable = NA_real_,
      mean_dist = NA_real_, median_dist = NA_real_,
      q25 = NA_real_, q75 = NA_real_,
      sample_dist = list(numeric(0))
    ))
  }

  # SAFE weights retrieval (no direct E(g_comp)[[...]] indexing)
  if (!is.null(weights_attr) &&
      igraph::ecount(g_comp) > 0 &&
      weights_attr %in% igraph::edge_attr_names(g_comp)) {
    w <- igraph::edge_attr(g_comp, weights_attr)
  } else {
    w <- NA  # fall back to unweighted hops
  }

  D <- igraph::distances(g_comp, v = A, to = B, weights = w, mode = "all")

  n_self_removed <- 0L
  if (drop_self_pairs && identical(class_a, class_b)) {
    ra <- rownames(D); cb <- colnames(D)
    common <- intersect(ra, cb)
    if (length(common)) {
      D[cbind(match(common, ra), match(common, cb))] <- NA_real_
      n_self_removed <- length(common)
    }
  }

  Dfin <- as.numeric(D[is.finite(D)])
  total_pairs_nominal <- length(A) * length(B) - n_self_removed
  nfin <- length(Dfin)
  frac_reach <- if (total_pairs_nominal > 0) nfin / total_pairs_nominal else NA_real_

  k <- min(sample_k, nfin)
  Dsample <- if (k > 0) sample(Dfin, k) else numeric(0)

  tibble::tibble(
    class_a = class_a,
    class_b = class_b,
    comp_size = igraph::vcount(g_comp),
    nA = length(A),
    nB = length(B),
    n_pairs = nfin,
    n_self_removed = n_self_removed,
    frac_reachable = frac_reach,
    mean_dist = if (nfin) mean(Dfin) else NA_real_,
    median_dist = if (nfin) stats::median(Dfin) else NA_real_,
    q25 = if (nfin) stats::quantile(Dfin, 0.25) else NA_real_,
    q75 = if (nfin) stats::quantile(Dfin, 0.75) else NA_real_,
    sample_dist = list(Dsample)
  )
}
```

```r
lst = lst_dat_region_annot_hic_import
idx = "fcc_astarr_macs_input_overlap"
dat = lst[[idx]]

g <- graph_from_data_frame(
  d = dat %>% select(Region_1, Region_2, weight = Loop_Score_Sum),
  directed = FALSE
)

# --- 0) Set length (you already did) ---
E(g)$len <- 1 / E(g)$weight

# --- 1) Component sizes & choose components to keep ---
comp <- components(g)
comp_sizes <- tibble(comp_id = seq_along(comp$csize), size = comp$csize)

min_size <- 50     # try 25, 50, 100 in sensitivity
keep_ids <- comp_sizes %>% filter(size >= min_size) %>% pull(comp_id)

# --- 2) Node table in this graph ---
node_df <- g_tbl %>%
  as_tibble() %>%
  select(name, Group, TSS_Proximity, Direction_Assay)

class_col <- "Group"  # or "TSS_Proximity"
classes <- sort(unique(node_df[[class_col]]))
class_pairs <- expand_grid(class_a = classes, class_b = classes)

min_per_class <- 3

# --- 3) Loop over kept components with per-class presence filter ---
per_comp_results <- map_dfr(keep_ids, function(cid){
  verts <- V(g)[comp$membership == cid]
  gC <- induced_subgraph(g, verts)

  nodeC <- node_df %>% filter(name %in% V(gC)$name)
  if (vcount(gC) < 2L) return(tibble())

  # which classes have enough nodes in this component?
  valid_classes <- nodeC %>%
    count(!!rlang::sym(class_col)) %>%
    filter(n >= min_per_class) %>%
    pull(!!rlang::sym(class_col))

  if (!length(valid_classes)) return(tibble())

  class_pairs_filt <- class_pairs %>%
    filter(class_a %in% valid_classes, class_b %in% valid_classes)

  map_dfr(seq_len(nrow(class_pairs_filt)), function(i){
    comp_pair_summary(
      g_comp = gC,
      node_tbl_comp = nodeC,
      class_a = class_pairs_filt$class_a[i],
      class_b = class_pairs_filt$class_b[i],
      class_col = class_col,
      n_per_class = Inf,
      weights_attr = "len",
      sample_k = 10000,
      drop_self_pairs = TRUE
    )
  }) %>%
    mutate(component_id = cid, comp_size = vcount(gC))
})

# --- 4) Aggregate across components (pair-weighted) + true global medians ---
class_dist_summary <- per_comp_results %>%
  group_by(class_a, class_b) %>%
  summarize(
    comps_with_pairs = sum(n_pairs > 0),
    total_pairs      = sum(n_pairs),
    mean_dist        = if (sum(n_pairs) > 0)
                         weighted.mean(mean_dist, w = n_pairs, na.rm = TRUE) else NA_real_,
    median_of_medians = median(median_dist[is.finite(median_dist)], na.rm = TRUE),
    overall_frac_reachable = if (n() > 0) mean(frac_reachable, na.rm = TRUE) else NA_real_,
    pooled_sample = list(unlist(sample_dist, use.names = FALSE)),
    .groups = "drop"
  ) %>%
  mutate(global_median = map_dbl(pooled_sample, ~ if (length(.x)) median(.x) else NA_real_))
```
