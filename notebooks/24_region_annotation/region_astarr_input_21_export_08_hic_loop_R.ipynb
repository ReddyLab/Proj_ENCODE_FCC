## Get region pair

**Helper function**

```r
fun_parse_region = function(txt_region){
    res = strcapture(
        "^(chr[^:]+):(\\d+)-(\\d+)$", 
        txt_region,
        proto = list(chrom = character(), start = integer(), end = integer())
    )
    return(res)
}
```

```r
fun_compare_region_order = function(txt_region_1, txt_region_2){
    dat_region_1 = fun_parse_region(txt_region_1)
    dat_region_2 = fun_parse_region(txt_region_2)
    
    vec_txt_chrom    = c(paste0("chr", 1:22), "chrX", "chrY", "chrM", "chrMT")
    idx_chrom_name_1 = as.integer(factor(dat_region_1$chrom, levels = vec_txt_chrom, ordered = TRUE))
    idx_chrom_name_2 = as.integer(factor(dat_region_2$chrom, levels = vec_txt_chrom, ordered = TRUE))
    
    is_left2right =
        (idx_chrom_name_1 <  idx_chrom_name_2) |
        (idx_chrom_name_1 == idx_chrom_name_2  &
            (dat_region_1$start < dat_region_2$start |
                (dat_region_1$start == dat_region_2$start &
                    dat_region_1$end <= dat_region_2$end
                )
            )
        )
    return(is_left2right)
}
```

**Get the region pair by loop**

```r
lst = lst_dat_region_annot_loop_anchor
lst = lapply(lst, function(dat){
    ### split regions by both sides of loop
    tmp = split(dat, dat$Label)
    
    ### rename the region intersect with loop anchor A
    vec_txt_cname1 = c("Chrom",   "ChromStart",   "ChromEnd",   "Region",   "Loop")
    vec_txt_cname2 = c("Chrom_1", "ChromStart_1", "ChromEnd_1", "Region_1", "Loop")
    vec_txt_loopup = setNames(vec_txt_cname1, vec_txt_cname2)
    
    dat = tmp$Loop_A
    dat = dat %>% 
        dplyr::select(all_of(vec_txt_cname1)) %>% 
        dplyr::rename(all_of(vec_txt_loopup))
    dat_loop_A = dat

    ### rename the region intersect with loop anchor B
    vec_txt_cname1 = c("Chrom",   "ChromStart",   "ChromEnd",   "Region",   "Loop")
    vec_txt_cname2 = c("Chrom_2", "ChromStart_2", "ChromEnd_2", "Region_2", "Loop")
    vec_txt_loopup = setNames(vec_txt_cname1, vec_txt_cname2)
    
    dat = tmp$Loop_B
    dat = dat %>% 
        dplyr::select(all_of(vec_txt_cname1)) %>% 
        dplyr::rename(all_of(vec_txt_loopup))
    dat_loop_B = dat
    
    ### link regions by joining loop anchor A and B intersected regions
    dat = dplyr::inner_join(dat_loop_A, dat_loop_B, by = "Loop", relationship = "many-to-many") 
    dat = dat %>% dplyr::distinct()
    
    ### check region order
    dat = dat %>% dplyr::mutate(is_Left2Right = fun_compare_region_order(Region_1, Region_2))

    ### arrange columns
    vec = c(
        "Chrom_1", "ChromStart_1", "ChromEnd_1", "Region_1",
        "Chrom_2", "ChromStart_2", "ChromEnd_2", "Region_2", 
        "Loop", 
        "is_Left2Right"
    )
    dat = dat %>% dplyr::select(all_of(vec))
    return(dat)
})

### assign and show
lst_dat_region_annot_loop_edge = lst

res = lapply(lst, dim)
print(res)

dat = lst[[1]]
fun_display_table(head(dat, 3))
```

**Check region order**

As all the regions are sorted by genome coordinate, there is no need to swap and prevent the possible duplication pair.

```r
lst = lst_dat_region_annot_loop_edge
lst = lapply(lst, function(dat){
    res = table(dat$is_Left2Right, dnn="is_Left2Right")
    dat = as.data.frame(res)
    return(dat)
})

dat = bind_rows(lst, .id = "Folder")
dat = dat %>% tidyr::spread(Folder, Freq)
fun_display_table(dat)
```
