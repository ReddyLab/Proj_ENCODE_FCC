```r
### init
dat = dat_hic_loop_arrange

### standardize loop scores
vec_num_score = dat$Score_Log2OE

vec = vec_num_score
num_med = median(vec, na.rm = TRUE)
vec = abs(vec - num_med)
num_mad = median(vec, na.rm = TRUE) * 1.4826

vec = vec_num_score
obj_dens = density(vec, na.rm = TRUE)
num_mode = obj_dens$x[which.max(obj_dens$y)]

vec = vec_num_score
num_mean = mean(vec, na.rm = TRUE)
num_std  = sd(vec, na.rm = TRUE)

dat = dat %>% dplyr::mutate(Zcore_Log2OE_by_Mean   = (Score_Log2OE - num_mean) / num_std)
dat = dat %>% dplyr::mutate(Zcore_Log2OE_by_Median = (Score_Log2OE - num_med)  / num_mad)
dat = dat %>% dplyr::mutate(Zcore_Log2OE_by_Mode   = (Score_Log2OE - num_mode) / num_mad)

dat_hic_loop_score_global = dat
print(dim(dat))
fun_display_table(head(dat, 3))
```

**distance-stratified normalization**

```r
### init
dat = dat_hic_loop_score_global

### Distance bins (log-spaced)
vec_num_breaks = c(3.5, 4, 4.5, 5, 5.5, 6, 6.5, 7)
dat = dat %>% dplyr::mutate(
    Dist_Log10 = log10(Distance),
    Dist_Bin   = cut(Dist_Log10, breaks = vec_num_breaks, right = FALSE)
)

### Build a stratified z: (score - mode_bin) / MAD_bin
dat = dat %>%
    dplyr::group_by(Dist_Bin) %>%
    dplyr::mutate(
        .vec  = Score_Log2OE,
        .med  = median(.vec, na.rm = TRUE),
        .madS = median(abs(.vec - .med), na.rm = TRUE) * 1.4826,
        .mode = { v <- .vec[is.finite(.vec)]; d <- density(v); d$x[which.max(d$y)] },
        Zcore_Log2OE_by_DistStrat = (Score_Log2OE - .mode) / .madS
    ) %>%
    dplyr::ungroup() %>%
    dplyr::select(-.vec, -.med, -.madS, -.mode)

dat_hic_loop_score_strat = dat
print(dim(dat))
head(dat, 3)
```

```r
dat = dat_hic_loop_score_strat

# 1) center: loess of score ~ log10(distance)
fit_mu = loess(Score_Log2OE ~ Dist_Log10, data = dat, span = 0.3, degree = 1)

# 2) scale: loess of absolute residuals ~ x (robust)
resid   = dat$Score_Log2OE - predict(fit_mu, dat$Dist_Log10)
fit_sig = loess(abs(resid) ~ Dist_Log10, data = dat, span = 0.3, degree = 1)

# 3) smooth z
mu_hat  = predict(fit_mu,  dat$Dist_Log10)
sig_hat = predict(fit_sig, dat$Dist_Log10)

dat = dat %>% dplyr::mutate(
    Zcore_Log2OE_by_DistSmooth = (Score_Log2OE - mu_hat) / sig_hat
)

dat_hic_loop_score_smooth = dat
print(dim(dat))
head(dat, 3)
```

```r
### Replace step #0–#1 with quantile binning
# 0) Flag inter-chrom and compute log10 distance (for intra only)
dat0 <- dat_hic_loop_score %>%
  mutate(is_inter = Chrom_1 != Chrom_2,
         log10_dist = log10(pmax(Distance, 1L)))

# Build quantile breaks *from intra-chrom only*
xi <- dat0$log10_dist[!dat0$is_inter & is.finite(dat0$log10_dist)]
make_qbreaks <- function(x, probs = seq(0, 1, by = 0.1)) {
  brks <- quantile(x, probs = probs, na.rm = TRUE)
  brks <- unique(as.numeric(brks))                 # drop ties
  if (length(brks) < 3) brks <- pretty(x, n = 10)  # fallback if too few uniques
  brks
}
qbreaks <- make_qbreaks(xi)                         # e.g., deciles

# Factor levels for intra bins (ordered nicely)
levs_intra <- levels(cut(xi, breaks = qbreaks, include.lowest = TRUE, right = FALSE))

# 1) Assign bins: 'inter' for inter-chrom; quantile bins for intra-chrom
dat1 <- dat0 %>%
  mutate(
    dist_bin = if_else(
      is_inter, "inter",
      as.character(cut(log10_dist, breaks = qbreaks,
                       include.lowest = TRUE, right = FALSE))
    ),
    dist_bin = factor(dist_bin, levels = c("inter", levs_intra), ordered = TRUE)
  ) %>%
  select(-log10_dist)


### Keep your per-bin normalization (unchanged)
dat_strat <- dat1 %>%
  group_by(dist_bin) %>%
  mutate(
    .vec  = Score_Log2OE,
    .med  = median(.vec, na.rm = TRUE),
    .madS = median(abs(.vec - .med), na.rm = TRUE) * 1.4826,
    .mode = { v <- .vec[is.finite(.vec)]; d <- density(v); d$x[which.max(d$y)] },
    Z_byModeMAD_dist = (Score_Log2OE - .mode) / .madS
  ) %>%
  ungroup() %>%
  select(-.vec, -.med, -.madS, -.mode)

# 3) Diagnostic A: trend with distance (raw vs global z)
p_raw <- ggplot(dat_strat %>% filter(!is_inter), aes(x = log10(Distance), y = Score_Log2OE)) +
  geom_bin2d(bins = 120) + geom_smooth(method = "loess", se = FALSE) +
  labs(title = "Raw log2(O/E) vs log10(distance)")

p_gz  <- ggplot(dat_strat %>% filter(!is_inter), aes(x = log10(Distance), y = Zcore_Log2OE_by_Mode)) +
  geom_bin2d(bins = 120) + geom_smooth(method = "loess", se = FALSE) +
  labs(title = "Global z (mode/MAD) vs log10(distance)")

p_sz  <- ggplot(dat_strat %>% filter(!is_inter), aes(x = log10(Distance), y = Z_byModeMAD_dist)) +
  geom_bin2d(bins = 120) + geom_smooth(method = "loess", se = FALSE) +
  labs(title = "Distance-stratified z vs log10(distance)")

# print(p_raw); print(p_gz); print(p_sz)

# 4) Diagnostic B: per-bin center/scale
summ <- dat_strat %>%
  group_by(dist_bin) %>%
  summarise(
    n          = n(),
    med_raw    = median(Score_Log2OE, na.rm = TRUE),
    mad_raw    = median(abs(Score_Log2OE - median(Score_Log2OE, na.rm = TRUE)), na.rm = TRUE) * 1.4826,
    med_gz     = median(Zcore_Log2OE_by_Mode, na.rm = TRUE),
    mad_gz     = median(abs(Zcore_Log2OE_by_Mode - median(Zcore_Log2OE_by_Mode, na.rm = TRUE)), na.rm = TRUE) * 1.4826,
    med_sz     = median(Z_byModeMAD_dist, na.rm = TRUE),
    mad_sz     = median(abs(Z_byModeMAD_dist - median(Z_byModeMAD_dist, na.rm = TRUE)), na.rm = TRUE) * 1.4826,
    .groups = "drop"
  )

# RMSE of bin medians from 0, and of scales from 1
rmse <- function(x) sqrt(mean(x^2, na.rm = TRUE))
rmse_med_gz <- rmse(summ$med_gz);     rmse_med_sz <- rmse(summ$med_sz)
rmse_mad_gz <- rmse(summ$mad_gz - 1); rmse_mad_sz <- rmse(summ$mad_sz - 1)

cat("RMSE(bin median)  Global z:", rmse_med_gz, "  Stratified z:", rmse_med_sz, "\n")
cat("RMSE(bin MAD-1)   Global z:", rmse_mad_gz, "  Stratified z:", rmse_mad_sz, "\n")

# 5) Diagnostic C: correlation with distance (should drop toward 0)
intra <- dat_strat %>% filter(!is_inter, is.finite(Distance))
cor_raw <- cor(intra$Score_Log2OE, log10(intra$Distance), method = "spearman", use = "complete.obs")
cor_gz  <- cor(intra$Zcore_Log2OE_by_Mode, log10(intra$Distance), method = "spearman", use = "complete.obs")
cor_sz  <- cor(intra$Z_byModeMAD_dist,     log10(intra$Distance), method = "spearman", use = "complete.obs")

cat("Spearman rho vs log10(distance):  raw =", cor_raw, 
    " global_z =", cor_gz, " strat_z =", cor_sz, "\n")

### Update Diagnostic D (densities) for dynamic bins
library(forcats)
levs <- setdiff(levels(dat_strat$dist_bin), "inter")
k <- length(levs)
cut1 <- max(1, floor(k/3))
cut2 <- max(cut1 + 1, floor(2*k/3))

short_levs <- levs[1:cut1]
mid_levs   <- levs[(cut1+1):cut2]
long_levs  <- levs[(cut2+1):k]

mk_class <- function(bin) dplyr::case_when(
  bin %in% short_levs ~ "short",
  bin %in% mid_levs   ~ "mid",
  bin %in% long_levs  ~ "long",
  TRUE                ~ NA_character_
)

library(ggplot2)
ggplot(dat_strat %>% filter(!is_inter) %>%
                  mutate(class = mk_class(dist_bin)),
                aes(x = Zcore_Log2OE_by_Mode, colour = class)) +
  geom_density() + labs(title = "Global z by distance class")
print(p_gz)

ggplot(dat_strat %>% filter(!is_inter) %>%
                  mutate(class = mk_class(dist_bin)),
                aes(x = Z_byModeMAD_dist, colour = class)) +
  geom_density() + labs(title = "Distance-stratified z by distance class")
print(p_sz)
```
