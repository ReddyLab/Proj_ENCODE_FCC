```r
lst = lapply(VEC_TXT_FOLDER, function(txt_folder){
    ### set file directory
    txt_fdiry = file.path(FD_RES, "analysis_fcc_correlation", txt_folder)
    txt_fname = "region.coverage.score.concat.tsv"
    txt_fpath = file.path(txt_fdiry, txt_fname)

    ### read table
    dat = read_tsv(txt_fpath, show_col_types = FALSE)
    return(dat)
})

### assign and show
lst_dat_region_score_import = lst

res = lapply(lst, dim)
print(res)

dat = lst[[1]]
fun_display_table(head(dat, 3))
```

```r
lst = lapply(VEC_TXT_FOLDER, function(txt_folder){
    ### set file directory
    txt_fdiry = file.path(
        FD_RES, 
        "region_annotation", 
        txt_folder,
        "summary"
    )
    txt_fname = "region.annotation.fcc_crispr.concat.tsv"
    txt_fpath = file.path(txt_fdiry, txt_fname)

    ### read table
    dat = read_tsv(txt_fpath, show_col_types = FALSE)
    return(dat)
})

### assign and show
lst_dat_region_annot_import = lst

res = lapply(lst, dim)
print(res)

dat = lst[[1]]
fun_display_table(head(dat, 3))
```

```r
vec_txt_assay_crispr = c(
    "CRISPRi-HCRFF", 
    "CRISPRi-Growth", 
    "ENCODE-rE2G"
)

lst = lst_dat_region_score_import
lst = lapply(lst, function(dat){
    ###
    dat = dat %>% dplyr::filter(Assay_Group %in% vec_txt_assay_crispr)

    ###
    dat = dat %>%
        dplyr::mutate(
            Assay_Group_CRISPR = Assay_Group,
            Assay_Label_CRISPR = Assay_Label,
            Assay_Score_CRISPR = Score
        ) %>%
        dplyr::select(
            Chrom, ChromStart, ChromEnd, Region, 
            Assay_Group_CRISPR,
            Assay_Label_CRISPR,
            Assay_Score_CRISPR
        ) %>%
        dplyr::distinct()
    return(dat)
})

### assign and show
lst_dat_region_score_arrange = lst

res = lapply(lst, dim)
print(res)

dat = lst[[1]]
fun_display_table(head(dat, 3))
```

```r
lst = lst_dat_region_annot_import
lst = lapply(lst, function(dat){
    ###
    dat = dat %>%
        dplyr::mutate(
            Assay_Group_CRISPR = Group,
            Assay_Label_CRISPR = Label,
            Assay_Annot_CRISPR = "Screen:Signif"
        ) %>%
        dplyr::select(
            Chrom, ChromStart, ChromEnd, Region, 
            Assay_Group_CRISPR, 
            Assay_Label_CRISPR,
            Assay_Annot_CRISPR
        ) %>%
        dplyr::distinct()
    return(dat)
})

### assign and show
lst_dat_region_annot_arrange = lst

res = lapply(lst, dim)
print(res)

dat = lst[[1]]
fun_display_table(head(dat, 3))
```

```r
lst = lst_dat_region_annot_arrange
lst = lapply(lst, function(dat){
    res = table(
        dat$Assay_Group_CRISPR, 
        dat$Assay_Label_CRISPR,
        dnn = c("Group", "Label")
    )
    dat = as.data.frame(res)
    return(dat)
})
dat = bind_rows(lst, .id = "Folder")
dat = dat %>% 
    dplyr::filter(Freq != 0) %>% 
    tidyr::spread(Folder, Freq)
fun_display_table(dat)
```

```r
lst = lapply(VEC_TXT_FOLDER, function(txt_folder){
    ### extract table
    dat_region_score = lst_dat_region_score_arrange[[txt_folder]]
    dat_region_annot = lst_dat_region_annot_arrange[[txt_folder]]

    ###
    vec = c(
        "Chrom", "ChromStart", "ChromEnd", "Region", 
        "Assay_Group_CRISPR", 
        "Assay_Label_CRISPR"
    )
    dat = dplyr::left_join(dat_region_score, dat_region_annot, by = vec)
    dat = dat %>% 
        dplyr::mutate(
            Assay_Annot_CRISPR = ifelse(
                is.na(Assay_Annot_CRISPR),
                "Screen:NotSignif",
                Assay_Annot_CRISPR
            )
        ) %>%
        dplyr::distinct()
    return(dat)
})

### assign and show
lst_dat_region_merge = lst

res = lapply(lst, dim)
print(res)

dat = lst[[1]]
fun_display_table(head(dat, 3))
```

```r
lst = lst_dat_region_merge
lst = lapply(lst, function(dat){
    res = table(
        dat$Assay_Label_CRISPR,
        dat$Assay_Annot_CRISPR,
        dnn = c("Label", "Annot")
    )
    dat = as.data.frame(res)
    return(dat)
})
dat = bind_rows(lst, .id = "Folder")
dat = dat %>% tidyr::spread(Folder, Freq)
fun_display_table(dat)
```

```r
for (txt_folder in VEC_TXT_FOLDER){

    ### get table
    dat_region_annot_result = lst_dat_region_merge[[txt_folder]]
    
    ### set file directory
    txt_fdiry = file.path(
        FD_RES, 
        "analysis_fcc_comparison", 
        txt_folder
    )
    
    ### set file path
    txt_fname = "region.annotation.fcc_crispr_concat.screen.tsv"
    txt_fpath = file.path(txt_fdiry, txt_fname)

    ### write table
    dat = dat_region_annot_result
    write_tsv(dat, txt_fpath)
}
```
