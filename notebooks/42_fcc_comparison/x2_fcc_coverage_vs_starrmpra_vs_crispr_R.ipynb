## Import data

```r
### loop to import data
lst = lapply(VEC_TXT_FOLDER, function(txt_folder){
    ### set file directory
    txt_fdiry = file.path(FD_RES, "region_annotation", txt_folder, "summary")
    txt_fname = "region.annotation.fcc_starrmpra_junke.group.label.tsv"
    txt_fpath = file.path(txt_fdiry, txt_fname)

    ### read table
    dat = read_tsv(txt_fpath, show_col_types = FALSE)
    return(dat)
})

### assign and show
lst_dat_region_annot_fcc_starrmpra_group = lst

res = lapply(lst, dim)
print(res)

dat = lst[[1]]
fun_display_table(head(dat, 3))
```

```r
lst = lapply(VEC_TXT_FOLDER, function(txt_folder){
    ### set file directory
    txt_fdiry = file.path(
        FD_RES, 
        "analysis_fcc_comparison", 
        txt_folder
    )
    txt_fname = "region.annotation.fcc_crispr_concat.screen.tsv"
    txt_fpath = file.path(txt_fdiry, txt_fname)

    ### read table
    dat = read_tsv(txt_fpath, show_col_types = FALSE)
    return(dat)
})

### assign and show
lst_dat_region_annot_fcc_crispr_concat = lst

res = lapply(lst, dim)
print(res)

dat = lst[[1]]
fun_display_table(head(dat, 3))
```

## Merge

```r
lst = lapply(VEC_TXT_FOLDER, function(txt_folder){
    ###
    dat_region_fcc_starrmpra = lst_dat_region_annot_fcc_starrmpra_group[[txt_folder]]
    dat_region_fcc_crispr    = lst_dat_region_annot_fcc_crispr_concat[[txt_folder]]
    
    ### read table
    vec = c("Chrom", "ChromStart", "ChromEnd", "Region")
    dat = dplyr::inner_join(
        dat_region_fcc_starrmpra,
        dat_region_fcc_crispr,
        by = vec,
        relationship ="many-to-many"
    )
    return(dat)
})

### assign and show
lst_dat_region_annot_fcc_merge = lst

res = lapply(lst, dim)
print(res)

dat = lst[[1]]
fun_display_table(head(dat, 3))
```

```r
for (txt_folder in VEC_TXT_FOLDER){

    ### get table
    dat_region_annot_result = lst_dat_region_annot_fcc_merge[[txt_folder]]
    
    ### set file directory
    txt_fdiry = file.path(
        FD_RES, 
        "analysis_fcc_comparison", 
        txt_folder
    )
    
    ### set file path
    txt_fname = "region.annotation.fcc_starrmpra_group.coverage.crispri.v2.tsv"
    txt_fpath = file.path(txt_fdiry, txt_fname)

    ### write table
    dat = dat_region_annot_result
    write_tsv(dat, txt_fpath)
}
```
